```python
import numpy as np
import pandas as pd
from scipy import sparse
from sklearn.cluster import KMeans
```

```python
def getLabels(indx_arr):
    return nodes_df_new.iloc[indx_arr, 2]
```

```python
def mismatch(labels_series):
   
    counts = labels_series.value_counts(ascending = True)
    #print("counts = ",counts)
    print("len = ",len(labels_series))
    if counts[0] == len(labels_series) or counts[1] == len(labels_series) :
        return 0
    else:
        count_0 = counts[0]
        count_1 = counts[1]
        if count_0 > count_1:
            return round(count_1/len(labels_series),2)
        else:
            return round(count_0/len(labels_series),2)
```

```python

nodes = []
with open("homework1/data/nodes.txt") as f:
    lines = f.readlines()
    
for line in lines:
    
    line =line.rstrip()
    node = line.split("\t")
    nodes.append(node)

nodes_df = pd.DataFrame(nodes, columns = ['Index', 'Name', 'Label', 'Info'])
nodes_df['Index'] = nodes_df['Index'].astype(int)
nodes_df['Isolated'] = 'F'
```

```python
edges = []
with open("homework1/data/edges.txt") as f:
    lines = f.readlines()
    
for line in lines:
    
    line =line.rstrip()
    edge = line.split("\t")
    edges.append(edge)
    
edges_df = pd.DataFrame(edges, columns = ['source', 'target'])
```

```python
edges_df
edges_df['source'] = edges_df['source'].astype(int)
edges_df['target'] = edges_df['target'].astype(int)
```

```python
###Pre-process - remove isolated nodes
for i in range(len(nodes_df)):
    node_idx = nodes_df.iloc[i,0]
    if node_idx not in edges_df.values: #the node is not in edges df
        nodes_df.iloc[i,4] = 'T'
        
nodes_df_new = nodes_df[nodes_df['Isolated'] == 'F']

nodes_df_new = nodes_df_new.reset_index(drop=True)

for  i in range(len(nodes_df_new)):
    if nodes_df_new.iloc[i,0] != i+1:
        old_index = nodes_df_new.iloc[i,0]
        new_index = i+1
        nodes_df_new.iloc[i,0] = new_index
        edges_df.replace(old_index, new_index,inplace = True)
        
```

```python
adjs = np.zeros((nodes_df_new.shape[0], nodes_df_new.shape[0]))
```

```python
adjs.shape
```

```python
for ind in edges_df.index:
     row, col = edges_df['source'][ind], edges_df['target'][ind]
     adjs[row-1][col-1]=1
     adjs[col-1][row-1]=1
```

```python
adjs.shape
```

```python
#t = edges_df[edges_df["source"] == 2]

sA = sparse.csr_matrix(adjs)
```

```python
s= sA.sum(axis=0)
#s
```

```python
D = np.diag(s.A1)
#D
```

```python
L=D-sA
```
